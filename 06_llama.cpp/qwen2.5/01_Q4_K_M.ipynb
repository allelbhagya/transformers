````python
PROMPT = """
Explain what a Transformer neural network is.
Explain self-attention, embeddings, and the role of the
Transformer layers. Keep the explanation technically accurate
but understandable to a machine learning student.
"""
````

````python
import time

from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:9931/v1",
    api_key="dummy"
)

start = time.perf_counter()

response = client.chat.completions.create(
    model="Qwen/Qwen2.5-0.5B-Instruct-GGUF:Q4_K_M",
    messages=[
        {"role": "user", "content": PROMPT}
    ],
    max_tokens=200,
)

end = time.perf_counter()
````

````python
print("\nresponse:")
print(response.choices[0].message.content)

print("\nmetrics:")
print(f"total request time: {end - start:.3f} seconds")
print(f"tokens generated: {response.usage.completion_tokens}")
print(f"prompt tokens: {response.usage.prompt_tokens}")
print(f"total tokens: {response.usage.total_tokens}")

generation_time = end - start

if response.usage.completion_tokens:
    print(
        f"approx tokens/sec: "
        f"{response.usage.completion_tokens / generation_time:.2f}"
    )
````
